# Introduction à la Data Science Moderne

## Qu'est-ce que la Data Science en 2025?

La data science moderne a considérablement évolué ces dernières années avec l'intégration des grands modèles de langage (LLMs), l'automatisation accrue et une focalisation sur des pratiques plus responsables.

### Tendances clés :
- **Intelligence artificielle générative** : Révolution des capacités d'analyse et de création
- **AutoML** : Démocratisation des processus de machine learning
- **MLOps** : Industrialisation des workflows de data science
- **IA éthique et responsable** : Considérations importantes sur les biais et la vie

```python
# Exemple simple d'analyse de données avec pandas
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Création d'un petit dataset d'exemple
data = {
    'Année': [2020, 2021, 2022, 2023, 2024, 2025],
    'Adoption_IA': [25, 32, 45, 58, 72, 85],
    'Croissance_Data_Science': [15, 22, 35, 42, 55, 65],
    'Intégration_LLM': [5, 10, 25, 45, 65, 80]
}

df = pd.DataFrame(data)
print("Aperçu des données:")
display(df)

# Visualisation simple
plt.figure(figsize=(10, 6))
sns.lineplot(data=df.melt(id_vars='Année',
                         value_vars=['Adoption_IA', 'Croissance_Data_Science', 'Intégration_LLM'],
                         var_name='Métrique', value_name='Pourcentage'))
plt.title('Évolution des tendances en Data Science (2020-2025)')
plt.ylabel('Pourcentage d\'adoption (%)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

## Le cycle de vie d'un projet de Data Science

Un projet de data science moderne suit généralement ces étapes:

1. **Définition du problème**
   - Compréhension des besoins métier
   - Formulation d'hypothèses testables

2. **Collecte et préparation des données**
   - Extraction des données de diverses sources
   - Nettoyage et prétraitement
   - Feature engineering

3. **Exploration et visualisation**
   - Analyse exploratoire des données (EDA)
   - Détection de patterns et d'anomalies

4. **Modélisation**
   - Sélection d'algorithmes appropriés
   - Entraînement et validation des modèles
   - Optimisation des hyperparamètres

5. **Évaluation et interprétation**
   - Mesure des performances
   - Explicabilité et interprétation
